# *Task 1: Sales Data Summary*
Before Run the code please ensure the resource file path
Data file given in resource folder

```python
import pandas as pd #pandas library to read csv file 

path =r"E:\HVLEARN\4 th assignment\CAR DETAILS FROM CAR DEKHO.csv"
# Load the car sales dataset from a local CSV file
car_sales_df = pd.read_csv(path)

# Group the data by 'fuel' column and count how many records are in each fuel type
fuel_type = car_sales_df.groupby("fuel").size()

# Filter only Petrol and Diesel fuel types
petrol_diesel_counts = fuel_type.loc[['Petrol', 'Diesel']]

# Calculate the total units sold (Petrol + Diesel)
Total_Units_sold = petrol_diesel_counts.sum()

# Calculate the absolute difference in units sold between Petrol and Diesel
Difference_Between_Categories = abs(petrol_diesel_counts['Petrol'] - petrol_diesel_counts['Diesel'])

# Calculate the ratio of Petrol to Diesel cars sold (rounded to 2 decimals)
ratio = round(petrol_diesel_counts['Petrol'] / petrol_diesel_counts['Diesel'], 2)

# Displaying the results in a clean format
print("Task 1: Sales Data Summary")
print("=== Sales Data Summary ===")
print(f"Category A (Petrol): {petrol_diesel_counts['Petrol']} units")
print(f"Category B (Diesel): {petrol_diesel_counts['Diesel']} units")
print(f"Total Units Sold: {Total_Units_sold}")
print(f"Difference Between Categories: {Difference_Between_Categories}")
print(f"Ratio of Category A to Category B: {ratio}")
print("===========================")
```

# *Task 13: Discount Calculation*

```python
print("Task 13: Discount Calculation")
print("=" * 55)

# Function to calculate final price after discount
def apply_discount(original_price, discount_percent):
    discount_amount = original_price * (discount_percent / 100)
    final_price = original_price - discount_amount
    return round(final_price, 2)

# List of products (Product Name, Original Price, Discount %)
products = [
    {"name": "Swift", "price": 600000, "discount": 5},
    {"name": "Baleno", "price": 670000, "discount": 7.5},
    {"name": "Celerio", "price": 490000, "discount": 10},
    {"name": "Ertiga", "price": 850000, "discount": 12},
    {"name": "WagonR", "price": 520000, "discount": 6}
]

# Function to apply discount
def apply_discount(price, discount_percent):
    return round(price - (price * discount_percent / 100), 2)

# Display discounted prices
for product in products:
    final_price = apply_discount(product["price"], product["discount"])
    print(f'{product["name"]:<10} - Original: ₹{product["price"]:,} | Discount: {product["discount"]}% → Final: ₹{final_price:,}')


print("=" * 55)
```

# *Task 14: Customer Feedback Sentiment Analysis*

```python
print("Task 14: Customer Feedback Sentiment Analysis")
print("=" * 60)

# Sample feedback list
feedback_list = [
    "The service was good and I am happy with the experience.",
    "I am disappointed by the delay in delivery.",
    "Great product, very satisfied!",
    "The quality is bad and I want a refund.",
    "It was okay, nothing special."
]

# Define keyword sets
positive_words = {"good", "happy", "great", "satisfied", "excellent", "awesome", "love"}
negative_words = {"bad", "disappointed", "poor", "terrible", "hate", "worst", "refund"}

# Analyze feedback
for idx, feedback in enumerate(feedback_list, start=1):
    feedback_lower = feedback.lower()  # Normalize
    sentiment = "Neutral"

    if any(word in feedback_lower for word in positive_words):
        sentiment = "Positive"
    if any(word in feedback_lower for word in negative_words):
        # If both positive and negative exist, we'll treat as Negative
        sentiment = "Negative"

    print(f"Feedback {idx}: {sentiment} → \"{feedback}\"")

print("=" * 60)
```

# *Task 17: Stock Replenishment Planning*

```python
print("Task 17: Stock Replenishment Planning")
print("=" * 60)

# Estimated average daily car sales
average_daily_car_sales = 5
reorder_threshold = average_daily_car_sales * 7  # 35 units

# List of car models and their stock
car_stock_list = [
    {"Model": "Swift", "Stock": 25},
    {"Model": "Baleno", "Stock": 40},
    {"Model": "Ertiga", "Stock": 30},
    {"Model": "Celerio", "Stock": 15},
    {"Model": "Fronx", "Stock": 45},
    {"Model": "Alto K10", "Stock": 20},
    {"Model": "Brezza", "Stock": 50},
    {"Model": "WagonR", "Stock": 33}
]

# Step 1: Check which cars need to be reordered
print(f"🚗 Reorder Threshold (7 days buffer): {reorder_threshold} units\n")
print("🔻 Car Models That Need Reordering:")

reorder_required = False
for car in car_stock_list:
    model = car["Model"]
    stock = car["Stock"]
    
    if stock < reorder_threshold:
        print(f" - {model:<10} | Current Stock: {stock} units")
        reorder_required = True

if not reorder_required:
    print("✅ All car models have sufficient stock.")

print("=" * 60)
```

# *Task 19: Simple Sales Forecasting*

```python
def main():
    print("Task 19: Simple Sales Forecasting")
    print("=" * 60)

    # Sales for last 3 months
    sales_last_3_months = [20000, 22000, 25000]

    # Calculate simple average
    forecast = sum(sales_last_3_months) / len(sales_last_3_months)

    # Display monthly sales
    print("📊 Last 3 Months Sales:")
    for i, sale in enumerate(sales_last_3_months, start=1):
        print(f" - Month {i}: ₹{sale:,.2f}")

    # Step 3: Display forecasted value
    print(f"\n🔮 Forecasted Sales for Next Month: ₹{forecast:,.2f}")
    print("=" * 60)

# Run the forecasting function
main()

```

# *Task 20: Customer Loyalty Points Calculator*

```python
print("Task 20: Customer Loyalty Points Calculator")
print("=" * 60)

# Step 1: List of customers and their total purchase amounts
customers = {
    "Arjun": 3200,
    "Meera": 7500,
    "Rajeev": 12000,
    "Divya": 9800,
    "Fatima": 4000,
    "Sohail": 15000
}

# Step 2: Define function to calculate loyalty points
def calculate_points(amount):
    if amount >= 10000:
        return int(amount * 2)
    elif amount >= 5000:
        return int(amount * 1.5)
    else:
        return int(amount * 1)

# Step 3: Print loyalty points for each customer
print("🧾 Customer Loyalty Points:\n")
for name, purchase in customers.items():
    points = calculate_points(purchase)
    print(f" - {name:<10} | Purchase: ₹{purchase:>6,} | Loyalty Points: {points}")

print("=" * 60)
```
